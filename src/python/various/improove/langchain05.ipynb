```python
import os
from dotenv import load_dotenv, find_dotenv
result = load_dotenv(find_dotenv()) # read local .env file
```

```python
from typing import List
from pydantic import BaseModel, Field

class ChangeTaskTitle(BaseModel):
    """Change the title of a task"""
    task_id: int = Field(description="The task id")
    title: str = Field(escription="The new title of the task")
```

```python
from langchain_openai  import AzureChatOpenAI
llm = AzureChatOpenAI(
    openai_api_version="2023-09-01-preview",
    deployment_name="GPT-4o-mini", #Deployment name
    azure_endpoint=os.environ["OPENAI_API_BASE"],
    model_name="GPT-4o-mini"
)
```

```python
tools = [ChangeTaskTitle]
llm_with_tools = llm.bind_tools(tools)
```

```python
result = llm_with_tools.invoke("Change Title of Task_645 to 'new title'")
print (result)
```

```python
result = llm_with_tools.invoke("Who is leonardo da vinci")
print (result)
```

```python
tools = [ChangeTaskTitle]
llm_with_forced_tools = llm.bind_tools(tools, tool_choice="ChangeTaskTitle")
```

```python
result = llm_with_forced_tools.invoke("Who is leonardo da vinci")
print(result)
```

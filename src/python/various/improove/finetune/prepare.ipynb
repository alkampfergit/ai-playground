```python
def prepare_jsonl(fileName):
    import json, os

# Load the data from data.json
    with open(fileName, 'r') as file:
        data = json.load(file)

    # Prepare the JSONL data
    jsonl_data = []
    for item in data:
        record = {
            "messages": [
                {"role": "system", "content": "Network Barbossa is an assistant that is expert in networking"},
                {"role": "user", "content": item["question"]},
                {"role": "assistant", "content": item["answer"]}
            ]
        }
        jsonl_data.append(record)

    # Write the JSONL data to datal.json
    output_file = fileName.replace('.json', '.jsonl')

    # Check if the output file already exists and delete it
    if os.path.exists(output_file):
        os.remove(output_file)

    with open(output_file, 'w') as file:
        for record in jsonl_data:
            file.write(json.dumps(record) + '\n')
```

```python
prepare_jsonl("validation.json")    
prepare_jsonl("data.json")    
```

```python
import os
from dotenv import load_dotenv, find_dotenv
result = load_dotenv(find_dotenv()) # read local .env file
```

```python
from typing import List
from pydantic import BaseModel, Field

class ChangeTaskTitle(BaseModel):
    """Change the title of a task"""
    task_id: str = Field(description="The task id")
    title: str = Field(escription="The new title of the task")
```

```python
from langchain_core.utils.function_calling import convert_to_openai_function
change_title_function = convert_to_openai_function(ChangeTaskTitle)
```

```python
from pprint import pprint
pprint(change_title_function)
```

```python
from langchain_openai  import AzureChatOpenAI
llm = AzureChatOpenAI(
    openai_api_version="2023-09-01-preview",
    deployment_name="GPT-4o-mini", #Deployment name
    azure_endpoint=os.environ["OPENAI_API_BASE"],
    model_name="GPT-4o-mini"
)

result =llm.invoke("Change the title of Task_125 to 'hey there'",
    functions = [change_title_function])
print(result)
```

```python
bound_call = llm.bind(functions=[change_title_function])
result = bound_call.invoke("Change Title of Task_645 to 'new title'")
pprint (result)
```

```python
result = bound_call.invoke("Who is Leonardo Da Vinci?")
pprint (result)
```
